```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

DATA_PATH = Path("../data/processed/counterpress_features.parquet")
TABLES_PATH = Path("../results/tables")
df = pd.read_parquet(DATA_PATH)
print(f"Loaded feature dataset: {df.shape[0]} rows, {df.shape[1]} columns")
```

## 1. Outcome Prevalence in Primary Cohort

```
print("Controlled Regain 5s:")
print(df["regain_5s_controlled"].value_counts(normalize=True).round(4))
print("
Dangerous Escape 15s:")
print(df["dangerous_escape_15s"].value_counts(normalize=True).round(4))
print("
Three-Class Tactical Outcome:")
print(df["episode_outcome_3class"].value_counts(normalize=True).round(4))
```

## 2. Descriptive Outcome Tables
Binned rates across local advantage, nearest player distances, forward advantage, and pitch zones.

```
desc_df = pd.read_csv(TABLES_PATH / "descriptive_outcome_tables.csv")
print(desc_df.head(20).to_string(index=False))
```

## 3. Local Numerical Advantage and Regain Rates

```
fig, ax = plt.subplots(figsize=(8, 4.5))
regain_adv = desc_df[(desc_df["grouping_variable"] == "numerical_advantage_10") & (desc_df["outcome"] == "regain_5s")]
ax.bar(regain_adv["bin_or_category"], regain_adv["outcome_rate"], color="#3182bd", edgecolor="#08519c")
ax.set_title("Regain within 5s Rate by Numerical Advantage at 10 Units")
ax.set_ylabel("Observed Regain Rate")
ax.axhline(0.3465, color="red", linestyle="--", label="Cohort Baseline (34.65%)")
ax.legend()
plt.show()
```
